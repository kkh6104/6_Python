문제 5. 플랫폼별 전 세계 판매량

플랫폼(Platform) 별로 Global_Sales를 합산하고, 판매량이 높은 순으로 TOP 10을 조회하시오.

```python
print(f"플랫폼 별 Global_Sales")
print(f"{df.groupby('Platform')['Global_Sales'].sum().sort_values(ascending=False).head(10)}")
```

문제 6. 가장 많이 판매된 장르

장르별 Global_Sales 총합을 구해 가장 높은 장르를 찾으시오.

```python
print(f"가장 인기 있는 장르 : {df.groupby('Genre')['Global_Sales'].sum().idxmax()} ({df.groupby('Genre')['Global_Sales'].sum().max()})")
```

문제 7. Publisher 별 평균 판매량

Publisher 별 평균 Global_Sales를 구하고, 상위 10개만 출력하시오.

```python
print(f"퍼블리셔 별 평균 Global_Sales")
print(f"{df.groupby('Publisher')['Global_Sales'].mean().sort_values(ascending=False).head(10)}")
```

문제 9. 국가별로 인기 있는 장르

NA / EU / JP 각각 판매량이 가장 높은 장르를 구하시오.

```python
for n in ['NA', 'EU', 'JP']:
    ns = df.groupby('Genre')[f'{n}_Sales'].sum()
    # print(ns)
    print(f"{n} 에서 판매량이 가장 높은 장르: {ns.idxmax()}")
```
